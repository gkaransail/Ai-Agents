loop to check all the pdf files in the folder

```python
import os
pdfs = []

for root, dirs, files in os.walk('/content/sample_data/PDFS/'):
  print(root, dirs, files)
  for file in files:
    if file.endswith('.pdf'):
      pdfs.append(os.path.join(root, file))



#Combines all the root director and files into a single
```

```python
pdfs
```

```python
docs =[]

for pdf in pdfs:
  loader = PyMuPDFLoader(pdf)
  pages = loader.load()
  docs.extend(pages)
```

```python
len(docs)
```

Document Chunking

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=100
)


chunks = text_splitter.split_documents(docs)
```

**Connecting to FAISS Vector database**

```python
index = faiss.IndexFlatL2(len(single_vector))
index
```

Checks the total item in the vector **db**

```python
index.ntotal, index.d
```

```python
vector_store = FAISS(
    embedding_function=embeddings,
    index=index,
    docstore=InMemoryDocstore(),
    index_to_docstore_id={}
)
```

```python
# help(vector_store)
```

 **Ingest the chunks as embeddings into Vector**

```python
ids = vector_store.add_documents(documents=chunks)
```

```python
len(ids)
```

```python
vector_store.index_to_docstore_id
len(vector_store.index_to_docstore_id)
```

**Store Vector Database**

```python
db_name = "health_supplements"
vector_store.save_local(db_name)
```

**Load Vector Database**

```python
 new_vector_store = FAISS.load_local(db_name,embeddings=embeddings,allow_dangerous_deserialization=True)
```

```python
len(new_vector_store.index_to_docstore_id)
```

**Retreival**

```python
question = "what is used to gain muscle mass?"

vector_store.search(query=question,search_type="similarity")


for doc in docs:
  print(doc.page_content)
  print("\n\n")
```

```python
retriever = vector_store.as_retriever(search_type="mmr",search_kwargs={"k":3,'fetch_k':100,'lambda_mult':1})
```

```python
docs = retriever.invoke(question)
for doc in docs:
  print(doc.page_content)
  print("\n\n")
```
